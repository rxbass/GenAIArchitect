```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
import os

load_dotenv(r"D:\ai\GenAIArchitect\.env", override=True)


prompt = ChatPromptTemplate.from_template(
    "You are a product description generator. "
    "Write a creative and engaging product description "
    "for the following product: {product}"
)

model = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0.9
)

chain = prompt | model | StrOutputParser()

result = chain.invoke({
    "product": "iPhone"
})

print(result)
```

```python
import langsmith as ls

with ls.tracing_context(
    enabled=True,
    project_name="langchain-assignment"
):
    result = chain.invoke({"product": "iPhone"})

print(result)
```
