```python
#text splitter

from langchain_text_splitters import RecursiveCharacterTextSplitter


text = "This is a sample text for data ingestion. " \
"It contains multiple sentences and can be used to demonstrate how to process text data in Python. " \
"The goal is to extract meaningful information from the text and perform various operations " \
"such as tokenization, sentiment analysis, and more."

text_length = len(text)

print("Length of the text:", text_length)

print("Splitting the text into chunks...")
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=80, 
    chunk_overlap=30,
    length_function=len
)


for i, chunk in enumerate(text_splitter.split_text(text)):
    print(f"Chunk {i+1}: {chunk} - Length: {len(chunk)}")  
```

```python
#text splitter for text document

from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter


text_loader = TextLoader("sample_text.txt")
documents = text_loader.load()

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=80,
    chunk_overlap=30,
    length_function=len
)
chunks = text_splitter.split_documents(documents);

for i, chunk in enumerate(chunks):
    print(f"Chunk {i+1}: {chunk.page_content} - Length: {len(chunk.page_content)}")

```
