```python
# create a sample file to load data from text 

text = "This is a sample text for data ingestion. It contains multiple sentences and can be used to demonstrate how to process text data in Python. The goal is to extract meaningful information from the text and perform various operations such as tokenization, sentiment analysis, and more."

with open("sample_text.txt", "w") as file:
    file.write(text)

print("Sample text file 'sample_text.txt' created successfully.")
```

```python
# Load the text with TextLoader

from langchain_community.document_loaders import TextLoader

loader = TextLoader("sample_text.txt")
documents = loader.load()

print("Number of documents loaded:", len(documents))
print("Content of the first document:")
print(documents[0].page_content)
```

```python
#load Pdf document with PyPDFLoader

from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("EULA-1.pdf")
documents = loader.load()
print("Number of documents loaded:", len(documents))
print("Content of the first document:")
print(documents[0].page_content)
```
