```python
#load open ai embedding model

from langchain_openai import OpenAIEmbeddings
from dotenv import load_dotenv
import os
load_dotenv()

api_key = os.getenv("OPENAI_API_KEY")

openai_embeddings = OpenAIEmbeddings(model="text-embedding-3-small", openai_api_key=api_key)

print("OpenAI Embeddings model loaded successfully.")
```

```python
#turn text into numbers
from sklearn.metrics.pairwise import cosine_similarity

text1 = "The cat is sleeping on the sofa."
text2 = "A cat is resting on the couch."
text3 = "A dog is playing in the yard."
embedding_vector1 = openai_embeddings.embed_query(text1)
embedding_vector2 = openai_embeddings.embed_query(text2)
embedding_vector3 = openai_embeddings.embed_query(text3)
print(f"Embedding vector for the first text: {embedding_vector1}")
print(f"Embedding vector for the second text: {embedding_vector2}")
print(f"Embedding vector for the third text: {embedding_vector3}")

#cosine similarity between the two embedding vectors
cosine_sim = cosine_similarity([embedding_vector1], [embedding_vector2])
similarity = cosine_sim[0][0]
print(f"Cosine similarity between the two texts text 1 and text 2: {similarity}")
cosine_sim2 = cosine_similarity([embedding_vector1], [embedding_vector3])
similarity2 = cosine_sim2[0][0]
print(f"Cosine similarity between the two texts text 1 and text 3: {similarity2}")
```
